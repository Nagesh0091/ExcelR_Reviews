## Google Reviews Data Collection

In this project, Google reviews were scraped for the **ExcelR Solutions** training institute located in **Pune, Maharashtra**. The institute has two branches in Pune:

- **Viman Nagar**
- **Baner**
- **Bangaluru (Marathahalli)**

To ensure comprehensive data collection, reviews were extracted from Google Maps using all available review sorting filters:

1. **Most Relevant**
2. **Newest**
3. **Highest Rating**
4. **Lowest Rating**

Each filter was scraped separately for both branches to capture the maximum number of available reviews. Since different filters may display overlapping or unique sets of reviews, this approach helped ensure a more complete dataset.

After collecting the data, the datasets obtained from all filters and both locations were **combined into a single dataset**. During the data preprocessing stage, **duplicate records were identified and removed** to ensure that each review appeared only once in the final dataset.

Finally, the cleaned and consolidated dataset was **exported as a single CSV file** for further analysis.
