```python
# The Google Maps review structure groups each review inside a single <div> container. 
# This container includes the reviewer’s name, rating, review text, and the time the review was posted. 
# By identifying this parent element and extracting the relevant child elements, 
# we can systematically collect all the review data for analysis.
```

```python
# Changing the filter to get the Newest, Highest ratings and Lowest ratings reviews.

# Sort

'''
<button class="S9kvJb " aria-label="Sort reviews" data-value="Sort"
'''

# Newest 


```

```python
from selenium.webdriver.common.by import By
import time

# Click Sort
sort_button = driver.find_element(By.XPATH,'//button[@aria-label="Sort reviews"]')
sort_button.click()

time.sleep(2)

# Click Newest
newest = driver.find_element(By.XPATH,'//div[@data-index="3"]') # changes index {"Newest":1,"Highest rating":2,:"Lowest rating":3}
newest.click()

time.sleep(3)
```

```python
last_height = 0
same_count = 0

while True:
    scrollable_div = driver.find_element(By.CSS_SELECTOR,"div.m6QErb.DxyBCb.kA9KIf.dS8AEf")

    driver.execute_script(
        "arguments[0].scrollTop = arguments[0].scrollHeight",
        scrollable_div
    )

    time.sleep(5)

    new_height = driver.execute_script(
        "return arguments[0].scrollHeight",
        scrollable_div
    )

    if new_height == last_height:
        same_count += 1
    else:
        same_count = 0

    if same_count >= 5:   # try 5 times before stopping
        break

    last_height = new_height
```

```python
review_parents = driver.find_elements(By.CSS_SELECTOR,"div.jftiEf")

for parent in review_parents:

    name = parent.find_element(By.CSS_SELECTOR,"div.d4r55").text
    stars = parent.find_element(By.CSS_SELECTOR,"span.kvMYJc").get_attribute("aria-label")

    try:
        review_text = parent.find_element(By.CSS_SELECTOR,"span.wiI7pd").text
    except:
        review_text = None

    time_since = parent.find_element(By.CSS_SELECTOR,"span.rsqaWe").text

    # Store in dataframe
    ExcelR_reviews.loc[len(ExcelR_reviews)] = [name, stars, review_text, time_since]
```
